# **Bar Plot**

```python
import numpy as np
```

```python
from matplotlib import pyplot as plt
```

```python
fruits={"Apple":5, "Banana":6,"Pineapple":8}
```

```python
print(fruits)
```

```python
type(fruits)
```

```python
names=list(fruits.keys())
values=list(fruits.values())
```

```python
plt.bar(names,values)
plt.show()
```

## **Adding titles and lables to the bar plot**

```python
plt.bar(names,values)
plt.title("Bar Plot")
plt.xlabel("Fruit")
plt.ylabel("Quantaty")
plt.show()
```

## **Horizontal barplot**

```python
plt.barh(names,values, color="purple")
plt.title("Bar Plot")
plt.xlabel("Fruit")
plt.ylabel("Quantaty")
plt.grid(True)
plt.show()
```
